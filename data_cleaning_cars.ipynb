```python
df['Sahiblər'].unique()
```

```python
def sahibler(x):
    if x == '4 və daha çox':
        return '4'
    else:
        return x
```

```python
df['Sahiblər'] = df['Sahiblər'].apply(sahibler)
df['Sahiblər'].unique()
```

```python
df['Sahiblər'] = pd.to_numeric(df['Sahiblər'], errors='coerce')
```

```python
df.info()
```

```python
df['Sahiblər'] = df['Sahiblər'].fillna(df['Sahiblər'].median())
```

```python
df.info()
```

```python
df['Yerlərin sayı'].unique()
```

```python
def yer(x):
    if x == '8+':
        return '8'
    else:
        return x
```

```python
df['Yerlərin sayı'] = df['Yerlərin sayı'].apply(yer)
df['Yerlərin sayı'].unique()
```

```python
df['Yerlərin sayı']=pd.to_numeric(df['Yerlərin sayı'], errors='coerce')
```

```python
df['Yerlərin sayı'] = df['Yerlərin sayı'].fillna(df['Yerlərin sayı'].median())
```

```python
exchange_rates = {
    'AZN': 1.00,
    '$':   1.70,
    '€':   1.98
}

df['price_azn'] = df.apply(
    lambda row: row['price_x'] * exchange_rates.get(row['currency_x'], 1.0),
    axis=1
)
```
